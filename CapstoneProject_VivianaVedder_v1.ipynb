## 1. Background and Roadmap

### 1.1 Background

**Cardiovascular diseases** (CVDs) are the leading cause of death. Of CVDs, congenital heart diseases are the most common congenital defects, with a prevalence of 1 in 100 live births. Despite the widespread knowledge that prenatal and postnatal drug exposure can lead to congenital abnormalities, the developmental toxicity of many FDA-approved drugs is rarely investigated. Therefore, to improve our understanding of drug side effects, we performed a high-content drug screen of **1,280 compounds** from the **Prestwick Drug Library** using zebrafish (*Danio rerio*) as an *in vivo* model for cardiovascular analyses. Zebrafish are a well-established model for CVDs and developmental toxicity. Here, we utilized the heart rate as a measurement under the assumption, that functional and developmental defects will result in an altered heartbeat. In our study, about 10.5% of the tested drugs significantly affected HR at a concentration of **20 µM** in zebrafish embryos at 2 days post-fertilization.

The data was obtained by collecting zebrafish eggs after 1 h of pair- or group-wise mating before over-night incubation. At 21 hours post-fertilization (hpf), the chorion (egg shell) of the embryos was removed (dechorionation). Then, at 24 hpf embryos were treated with compounds from the Prestwick Drug Library. Treatment was stopped at 48 hpf, and fish were mounted in orientation plates to visualize fluorescent hearts using the ACQUIFER imaging machine. Then quantitative data was obtained via analysis with pyHeart4Fish.

As previously described, this project utilizes data from the PhD project <a href="https://d-nb.info/1376022915" target="_blank" style="color:#66ccff; text-decoration:none;">

- Creating a cohesive table with clean, consistent data.
- Exploring the Drug Library used for this screening.
- Transforming numerous categorical columns into numerical data.
- Applying the "Horse-race"-Approach to identify the most suitable ML algorithm.
- Applying the best algorithm to the data.
- Testing the trained algorithm on new data (unsupervised learning).
- Assessing the performance.

#### 3.1.1 Changing Storage strings into numeric values

While testing standards define room temperature as 23°C, the chemical and thermodynamic standard for room temperature is 25°C. Therefore, 25°C will be applied here.

```python
prestw["Storage (powder form)"] = prestw["Storage (powder form)"].replace('\n', '', regex=True)
```

```python
print(prestw["Storage (powder form)"].value_counts())
```

```python
#Change Temperature labels to numeric values
temp_mapping = {
    'Store at room temperature': 25,
    'Store at RT': 25,
    'RT': 25,
    'Room Temp': 25,
    'Store at 2°C': 2,
    'Store at +2°C': 2,
    'Store at +4°C': 4,
    "store at +4°C": 4,
    'Store at 4°C': 4,
    "Store at 2-8°C": 4,
    "2-8°C": 4,
    'Store at -20°C': -20,
    "store at -20°C": -20,
    "Store at -20": -20,
    "Store -20°C": -20,
    "-20°C": -20,
    "Store below 0°C": -1,
    "Store at 0°C": -1
    }

prestw["Storage (powder form)"] = prestw["Storage (powder form)"].replace(temp_mapping)

prestw["Storage (powder form)"] = pd.to_numeric(prestw["Storage (powder form)"], errors='coerce')

print(prestw["Storage (powder form)"].value_counts())
```

#### 3.1.2 Cleaning and splitting Lit. Soluability column into 4 columns:

```python
#Removing descriptives from 'Lit. Solubility' column
prestw["Lit. Solubility"] = prestw["Lit. Solubility"].replace('hot', '', regex=True)
prestw["Lit. Solubility"] = prestw["Lit. Solubility"].replace('Hot', '', regex=True)
prestw["Lit. Solubility"] = prestw["Lit. Solubility"].replace('Dilute', '', regex=True)
prestw["Lit. Solubility"] = prestw["Lit. Solubility"].replace('dilute', '', regex=True)
```

```python
def get_smiles(name):
    """Fetches the isomeric SMILES for a given chemical name from PubChem."""
    try:
        # Search by name and take the first result
        results = pcp.get_compounds(name, 'name')
        # Returns the isomeric SMILES if found
        return results[0].isomeric_smiles if results else None
    except Exception:
        return None

# Applies the function to create a new column
# Note: PubChem limits requests to ~5 per second
schem_clean['Pub_smiles'] = schem['Chemical name'].apply(get_smiles) #schem is used as filters were applied previously

schem_clean[['Chemical name', 'SMILES', 'Pub_smiles']].head(10)
```

```python
#Creating a table of counts for combinations of Therapeutic Class and Treatment Effect

# Define your column sets
groups = ['TClass_1', 'TClass_2', 'TClass_3']
subgroups = ['Teffect_1', 'Teffect_2', 'Teffect_3', 'Teffect_4', 'Teffect_5', 'Teffect_6']

results = []

for g_col in groups:
    for s_col in subgroups:
        # Skips if either column is entirely NaN (all values missing)
        if prestw[g_col].isna().all() or prestw[s_col].isna().all():
            continue
            
        # Performs the count
        # groupby().size() automatically ignores rows where values are NaN
        counts = prestw.groupby([g_col, s_col]).size().reset_index(name='count')
        
        # Renames columns to a generic name for easy merging later
        counts.columns = ['group_value', 'subgroup_value', 'count']
        
        results.append(counts)

# Combines all valid pairs into one final table
if results:
    final_counts = pd.concat(results, ignore_index=True)
    # Reorders columns for better readability
    final_counts = final_counts[['group_value', 'subgroup_value', 'count']]
else:
    final_counts = pd.DataFrame() # Returns empty if no valid pairs found

print(final_counts)
```

Here optuna is applied, because it´s a smart hyperparameter optimization framework, moving beyond randomized search of a GridSearch. Its faster, as it prunes trials by killing bad trials early and uses bayesian intelligence (Tree-structured Parzen Estimator (TPE)).

```python
# Define Function
def objective(trial):
    """Runs trials on set parameters for set model"""
    params = {
        "max_depth": trial.suggest_int("max_depth", 6, 12),
        "gamma": trial.suggest_float("gamma", 1, 5),
        # ... weitere Parameter ...
    }
    model = XGBRegressor(**params, random_state=42, interaction_constraints=constraints, n_jobs=-1)
    score = cross_val_score(model, X_train, y_train, cv=5, scoring="r2").mean()
    return score

# Starting optimization 
study = optuna.create_study(direction="maximize")
study.optimize(objective, n_trials=50, show_progress_bar=True)

# Show results
print("Beste Parameter:", study.best_params)
print("Bester Score:", study.best_value)
```

The tanimoto score is a standard in chemoinformatics to quantify the structural similarity between two molecules. It will be applied here to assess differences between compounds of the two library and to see wether the models training can be applied to the Selleckchem library on a chemical level. (This part is included in preparation to include the chemical structures into the algorithm in the future.)

Tanimoto score calculation:

```python
# Double check: No NaN-values and all Strings
prestw_smiles_clean = prestw_clean_merged['smiles'].dropna().astype(str)
schem_smiles_clean = schem_clean['Pub_smiles'].dropna().astype(str)

# Converts SMILES-strings in RDKit-molecule objects
mols_prestw = [Chem.MolFromSmiles(smi) for smi in prestw_smiles_clean]
mols_schem = [Chem.MolFromSmiles(smi) for smi in schem_smiles_clean]

# Removes failed conversions (None-objects)
mols_prestw = [m for m in mols_prestw if m is not None]
mols_schem = [m for m in mols_schem if m is not None]
```

```python
# Radius 2, 2048 Bits (Standard ECFP4)
morgan_gen = rdFingerprintGenerator.GetMorganGenerator(radius=2, fpSize=2048)

# Generates ECFP4-Fingerprints (Bitvektors, length 2048)
fps_prestw = [morgan_gen.GetFingerprint(m) for m in mols_prestw]
fps_schem = [morgan_gen.GetFingerprint(m) for m in mols_schem]
```

```python
similarity_scores = []

# Iterates through every Selleckchem-molecule
for fp_sc in fps_schem:
    # Calculates similarity of this molecule to ALL Prestwick molecules
    similarities = DataStructs.BulkTanimotoSimilarity(fp_sc, fps_prestw)
    # Finds the highest score (Nearest Neighbor) for this SC molecule
    max_similarity = max(similarities)
    similarity_scores.append(max_similarity)

# Calculate the overall average
average_tanimoto = np.mean(similarity_scores)

print(f"Average Tanimoto Score (Selleckchem to Prestwick): {average_tanimoto:.3f}")
```

The Selleckchem dataset is predicted to contain more compounds with a higher z-score, probably also more significant data. Its shifted to the left. Therefore more compounds are predicted to reduce zebrafish heart rate.

Tanimoto score interpretation:
- **0.8–1.0**: Very similar or identical. Your green and orange curves should overlap almost perfectly.
- **0.5–0.8**: Moderately similar. This is a common range in drug research.
- **< 0.3**: Very different (Large Drift). If your value is in this range, it explains the strong shift in the KDE plot.

Here, the Tanimoto score shows, that the datasets as they are at the moment are moderately similar, which is to be expected comparing two different drug libraries. This also confirms, that the KDE shift observed is real and likely caused by differences in molecules. As previously shown, the model already recognized the importance of the molecular weight.

## 7. Conclusions

This capstone project demonstrates that z-score normalized heart rate of zebrafish can be predicted with machine learning (XGBR) with a 98% precision rate.

The **Q-Q plot** confirms that the majority of data points align closely with the 45° reference line, validating the model’s overall performance. While the **KDE plot** reveals slight skewness and minor deviations at the outermost edges, there are no severe heavy tails or extreme outliers impacting the data’s integrity. Given that real-world biological data rarely aligns perfectly with theoretical distributions, these slight deviations confirm the model’s practical suitability. Furthermore, the **Tanimoto score** establishes that predictions for the Selleckchem drug library are statistically grounded rather than random, confirming the model's reliability for machine learning inference.

These results indicate that the model effectively predicts the direction and, in most cases, the significance of heart rate alterations induced by chemical compounds in zebrafish. Consequently, the model can be used to prioritize compounds within the Selleckchem drug library for screening. This approach refines the drug discovery process, significantly reducing the time, cost, and number of animals required—directly upholding the **3R principles (Replace, Reduce, Refine)**.

Given that **Molecular Weight** was one of the key features for the XGBR model, future iterations should incorporate full molecular structures, as structurally similar compounds often share targets and off-target effects. In fact, removing the human heart rate phenotype from the training and test data resulted in a greater importance of the molecular weight. This pipeline is transferable to other screening facilities utilizing organoids or animal models. If experimental validation confirms these predictions, it should become a standard policy to utilize machine learning pre-screening to adhere to modern ethical and efficiency standards.

**Limitations and Future Work**

A primary limitation is that the current model operates exclusively on z-score normalized data; future research should apply this pipeline to raw data or log-transformed data to better manage outliers. Additionally, the Prestwick Chemical Library exhibits an inherent bias toward certain compound groups; training on more diverse, low-bias datasets would enhance generalizability.
